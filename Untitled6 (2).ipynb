```python
# =========================================================
# Section 0: Setup
# =========================================================
import os, random, urllib.request, zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

# ---- Reproducibility ----
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# ---- Device ----
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using compute device: {device}")
```

```python
# =========================================================
# Section 1: Download NASA C-MAPSS
# =========================================================
ZIP_PATH = "CMAPSSData.zip"
DATA_DIR = "cmapss_data"

if not os.path.exists(DATA_DIR):
    print("Downloading NASA C-MAPSS dataset...")
    url = "https://data.nasa.gov/docs/legacy/CMAPSSData.zip"
    urllib.request.urlretrieve(url, ZIP_PATH)
    with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
        zip_ref.extractall(DATA_DIR)
    print("Dataset downloaded and extracted.")
else:
    print("Dataset already present, skipping download.")
```

```python
# =========================================================
# Section 2: Preprocessing
# =========================================================
columns = ['unit_nr', 'time_cycle', 'setting_1', 'setting_2', 'setting_3'] \
        + [f's_{i}' for i in range(1, 22)]

RUL_CAP    = 125
SEQ_LEN    = 30

# ---- Load train ----
train_df = pd.read_csv(
    f"{DATA_DIR}/train_FD001.txt", sep=r'\s+', header=None, usecols=range(26)
)
train_df.columns = columns

# ---- Load test + ground-truth RUL ----
test_df = pd.read_csv(
    f"{DATA_DIR}/test_FD001.txt", sep=r'\s+', header=None, usecols=range(26)
)
test_df.columns = columns

rul_test = pd.read_csv(
    f"{DATA_DIR}/RUL_FD001.txt", sep=r'\s+', header=None
)[0].values.astype(np.float32)            # true RUL per engine (last cycle)

# ---- Compute capped RUL target for training ----
max_cycle = train_df.groupby('unit_nr')['time_cycle'].max().rename('max_cycle')
train_df = train_df.merge(max_cycle, on='unit_nr', how='left')
train_df['RUL'] = (train_df['max_cycle'] - train_df['time_cycle']).clip(upper=RUL_CAP)
train_df.drop(columns=['max_cycle'], inplace=True)

# ---- Drop constant / noise channels ----
drop_cols = ['setting_1', 'setting_2', 'setting_3',
             's_1', 's_5', 's_6', 's_10', 's_16', 's_18', 's_19']
feature_cols = [c for c in columns[5:] if c not in drop_cols]
print(f"Using {len(feature_cols)} features: {feature_cols}")

# ---- Fit scaler on TRAIN ONLY ----
scaler = StandardScaler()
scaler.fit(train_df[feature_cols])
train_df[feature_cols] = scaler.transform(train_df[feature_cols])
test_df[feature_cols]  = scaler.transform(test_df[feature_cols])

# ---- Rolling-mean features (degradation trend) ----
ROLL_WIN = 10
for c in list(feature_cols):
    train_df[f'{c}_roll'] = (
        train_df.groupby('unit_nr')[c].transform(
            lambda s: s.rolling(ROLL_WIN, min_periods=1).mean()
        )
    )
    test_df[f'{c}_roll'] = (
        test_df.groupby('unit_nr')[c].transform(
            lambda s: s.rolling(ROLL_WIN, min_periods=1).mean()
        )
    )

feature_cols = [c for c in train_df.columns
                if c not in ('unit_nr', 'time_cycle', 'RUL')]
print(f"Feature dimension after rolling stats: {len(feature_cols)}")
```

```python
# =========================================================
# Section 3: Sliding Window Sequences
# =========================================================
def gen_train_sequences(df, feature_cols, seq_length):
    """Window ends at cycle t; target = RUL at cycle t (aligned correctly)."""
    X, y = [], []
    for _, g in df.groupby('unit_nr'):
        if len(g) < seq_length:
            continue
        mat    = g[feature_cols].values.astype(np.float32)
        target = g['RUL'].values.astype(np.float32)
        for i in range(len(g) - seq_length + 1):            # <- fixed
            X.append(mat[i:i + seq_length])
            y.append(target[i + seq_length - 1])            # <- last cycle
    return np.asarray(X), np.asarray(y)


def gen_test_sequences(df, feature_cols, seq_length):
    """For test, take the final `seq_length` cycles per engine."""
    X = []
    for _, g in df.groupby('unit_nr'):
        mat = g[feature_cols].values.astype(np.float32)
        if len(mat) >= seq_length:
            X.append(mat[-seq_length:])
        else:                                               # pad left with first row
            pad = np.repeat(mat[:1], seq_length - len(mat), axis=0)
            X.append(np.concatenate([pad, mat], axis=0))
    return np.asarray(X)


X_all, y_all = gen_train_sequences(train_df, feature_cols, SEQ_LEN)
X_test       = gen_test_sequences (test_df,  feature_cols, SEQ_LEN)

print(f"X_all : {X_all.shape}    y_all : {y_all.shape}")
print(f"X_test: {X_test.shape}  y_test: {rul_test.shape}")

# ---- Train/Val split ----
X_tr, X_val, y_tr, y_val = train_test_split(
    X_all, y_all, test_size=0.15, random_state=SEED, shuffle=True
)
print(f"Train: {X_tr.shape} | Val: {X_val.shape}")
```

```python
# =========================================================
# Section 4: CNN-LSTM model (Dropout Removed)
# =========================================================
class RUL_CNNLSTM(nn.Module):
    def __init__(self, input_dim, hidden=64):
        super().__init__()
        # ---- local temporal feature extractor ----
        self.conv1 = nn.Conv1d(input_dim, 32, kernel_size=3, padding=1)
        self.bn1   = nn.BatchNorm1d(32)

        # ---- recurrent backbone ----
        self.lstm = nn.LSTM(
            input_size=32, hidden_size=hidden, num_layers=2,
            batch_first=True, dropout=0.0, bidirectional=False,
        )

        # ---- regression head ----
        self.norm = nn.LayerNorm(hidden)
        self.fc1  = nn.Linear(hidden, 32)
        self.fc2  = nn.Linear(32, 1)

    def forward(self, x):
        # x: (B, T, F) → Conv1d expects (B, F, T)
        x = x.transpose(1, 2)
        x = F.relu(self.bn1(self.conv1(x)))
        x = x.transpose(1, 2)                # back to (B, T, 32)

        out, _ = self.lstm(x)
        h = self.norm(out[:, -1, :])
        h = F.relu(self.fc1(h))
        return self.fc2(h).squeeze(-1)


model = RUL_CNNLSTM(input_dim=len(feature_cols)).to(device)
print(model)
print(f"Trainable params: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
```

```python
# =========================================================
# Section 5: Training
# =========================================================
def make_loader(X, y, batch_size, shuffle):
    ds = TensorDataset(
        torch.tensor(X, dtype=torch.float32),
        torch.tensor(y, dtype=torch.float32),
    )
    return DataLoader(ds, batch_size=batch_size, shuffle=shuffle,
                      num_workers=2, pin_memory=(device.type == 'cuda'))


train_loader = make_loader(X_tr,  y_tr,  batch_size=128, shuffle=True)
val_loader   = make_loader(X_val, y_val, batch_size=256, shuffle=False)

criterion  = nn.HuberLoss(delta=1.0)                    # robust to RUL outliers
optimizer  = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
scheduler  = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=3
)

EPOCHS       = 80
PATIENCE     = 8
best_val     = float('inf')
wait         = 0
train_hist   = []
val_hist     = []

for epoch in range(1, EPOCHS + 1):
    # ---------- train ----------
    model.train()
    tr_loss = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device, non_blocking=True), yb.to(device, non_blocking=True)
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        tr_loss += loss.item() * xb.size(0)
    tr_loss /= len(train_loader.dataset)

    # ---------- validate ----------
    model.eval()
    va_loss = 0.0
    with torch.no_grad():
        for xb, yb in val_loader:
            xb, yb = xb.to(device), yb.to(device)
            va_loss += criterion(model(xb), yb).item() * xb.size(0)
    va_loss /= len(val_loader.dataset)

    scheduler.step(va_loss)
    train_hist.append(tr_loss); val_hist.append(va_loss)

    lr_now = optimizer.param_groups[0]['lr']
    print(f"Epoch {epoch:3d}/{EPOCHS} | train={tr_loss:7.4f} | val={va_loss:7.4f} | lr={lr_now:.2e}")

    # ---------- early stopping ----------
    if va_loss < best_val - 1e-4:
        best_val, wait = va_loss, 0
        torch.save(model.state_dict(), "best_lstm.pt")
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping at epoch {epoch} (best val={best_val:.4f})")
            break

# ---- reload best weights ----
model.load_state_dict(torch.load("best_lstm.pt", map_location=device))
print("Loaded best checkpoint.")
```

```python
# =========================================================
# Section 6: Test-set evaluation
# =========================================================
def nasa_score(y_true, y_pred):
    """Asymmetric penalty used in the C-MAPSS literature."""
    d = y_pred - y_true
    return float(np.sum(np.where(d < 0, np.exp(-d / 13.0) - 1.0,
                                         np.exp( d / 10.0) - 1.0)))


@torch.no_grad()
def predict(arr, batch_size=256):
    model.eval()
    loader = DataLoader(
        torch.tensor(arr, dtype=torch.float32),
        batch_size=batch_size, shuffle=False,
    )
    out = []
    for xb in loader:
        out.append(model(xb.to(device)).cpu().numpy())
    return np.concatenate(out)


# ---- Test predictions ----
test_preds = predict(X_test)
test_preds = np.clip(test_preds, 0, RUL_CAP * 1.5)      # sanity clip

rmse_capped = np.sqrt(np.mean((test_preds - np.minimum(rul_test, RUL_CAP)) ** 2))
rmse_true   = np.sqrt(np.mean((test_preds - rul_test) ** 2))
score       = nasa_score(rul_test, test_preds)

print(f"Test RMSE (vs capped@{RUL_CAP}) : {rmse_capped:.3f}")
print(f"Test RMSE (vs raw true)         : {rmse_true:.3f}")
print(f"NASA score                      : {score:.2f}")
```

```python
# =========================================================
# Section 7: Visualizations
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# --- (a) learning curve ---
axes[0].plot(train_hist, label='Train (Huber)', linewidth=2)
axes[0].plot(val_hist,   label='Val   (Huber)', linewidth=2)
axes[0].set_title('Training / Validation Loss')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].legend(); axes[0].grid(True, ls=':', alpha=0.6)

# --- (b) test predictions vs true (sorted by true RUL) ---
order = np.argsort(rul_test)
axes[1].plot(rul_test[order],        label='True RUL',     color='C0', lw=2)
axes[1].plot(test_preds[order],      label='Predicted RUL', color='C1', ls='--', lw=2)
axes[1].set_title(f'FD001 Test Set — RMSE={rmse_capped:.2f}, NASA={score:.1f}')
axes[1].set_xlabel('Engine (sorted by true RUL)')
axes[1].set_ylabel('RUL (cycles)')
axes[1].legend(); axes[1].grid(True, ls=':', alpha=0.6)

plt.tight_layout(); plt.show()

# --- (c) scatter ---
plt.figure(figsize=(6, 6))
plt.scatter(rul_test, test_preds, alpha=0.7, edgecolor='k')
lim = max(rul_test.max(), test_preds.max()) + 5
plt.plot([0, lim], [0, lim], 'r--', lw=1.5, label='Ideal')
plt.xlabel('True RUL'); plt.ylabel('Predicted RUL')
plt.title('Predicted vs True RUL (Test)')
plt.legend(); plt.grid(True, ls=':', alpha=0.6)
plt.tight_layout(); plt.show()
```

```python
import torch

def save_model(model, filepath="rul_model_checkpoint.pth"):
    """Saves the trained model's state dictionary to disk."""
    torch.save(model.state_dict(), filepath)
    print(f"Model weights successfully saved to {filepath}")

def load_model(model, filepath="rul_model_checkpoint.pth", device='cpu'):
    """Loads saved model weights into the model architecture."""
    model.load_state_dict(torch.load(filepath, map_location=device))
    model.eval() # Set model to evaluation mode
    print(f"Model weights successfully loaded from {filepath}")
    return model
```

```python
import pickle
import torch

def save_model_pkl(model, filepath="rul_model.pkl"):
    """Saves the PyTorch model's state dictionary to disk using pickle."""
    # model.state_dict() returns a standard Python OrderedDict of weights,
    # which pickle can serialize easily.
    with open(filepath, 'wb') as f:
        pickle.dump(model.state_dict(), f)
    print(f"Model successfully saved to {filepath} via pickle!")


save_model_pkl(model, "turbofan_rul_model.pkl")
```
